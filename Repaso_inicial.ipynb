## Bloque 3
operadores

```python
# aritméticos +, -, *, /, //, %, **
print(1 + 1) #suma
print(2 - 1) #resta
print(2 * 3) #multiplicación
print(4 / 2) #división
print(13 // 2) #division con redondeo al menor (3,4 = 3, 3,7 = 3, 9,8 = 9)
print(13 % 2) #modulo, entrega el residuo
print(2 ** 3) #potencia
```

```python
# comparación ==, !=, <, >, <=, >=
print(2 == 2) #igualdad
print(2 != 2) #desigualdad
print(2 < 2) #menor que
print(2 > 2) #mayor que
print(2 <= 2) #menor o igual que
print(2 >= 2) #mayor o igual que
```

```python
# lógicos and, or, not
print(True and True) #and
print(True or False) #or
print(not True) #not
```

```python
# asignación compuesta +=, -=, *=, /=
x = 2
x += 2
print(x) # x = x + 2
x -= 2
print(x) # x = x - 2
x *= 2
print(x) # x = x * 2
x /= 2
print(x) # x = x / 2
```

## Bloque 4
- String (cadenas de texto)


```python
# se puede escribir texto con " " o ' '
print("Hola Python")
print('Hola Python')

# puedes concatener texto
print("Hola" + 'Juancito')

# formateo moderno
apellido = 3
print(f'Hola {apellido}') # para no concatenar con + y para evitar parsear a str() ya que print muestra solo str
```

```python
# podemos hacer llamados con indexación
nombre = ' Hola amigo' #cada letra y espacio se cuenta, comienza en 0.
print(nombre[2])
# slicing
print(nombre[0:2]) #extrae fragmento de texto según le indiques, letra posición 0 hasta letra posición 1.
print(nombre[0:-8]) #lo puedes hacer con numeros negativos, contando desde el final, depende de la situación.
```

```python
# métodos
print(nombre.upper()) #coloca todo en mayusculas
print(nombre.lower()) #coloca todo en minúscula
print(nombre.strip()) #elimina espacios que estén al inicio y final del string
print(nombre.replace("Hola","Hello")) #le indicas que palabras cambiar, o letra.
print(nombre.split()) #divide en sub cadenas
frase = ['hola', 'viejo', 'amigo']
print(" ".join(frase)) # inserta lo que tu le digas para univer string, en este caso una lista con palabras
```

## Bloque 7
- estructura de datos básica

```python
# método de listas [0, 1, 2, 3]
mapa = ['Lima','Los Angeles','Mexico','Durango']
print(mapa)
# agregar un dato al final
mapa.append("Chicago")
print(mapa)

# podemos agregar toda una lista a la lista
signos = ['aries','leo','libra']
mapa.extend(signos)
print(mapa)

# podemos agregar un dato en un lugar en especifico
mapa.insert(1, "4")
print(mapa)

# elimina el ultimo dato, o si colocas un numero elimna el dato de esa ubicación
mapa.pop()
print(mapa)

mapa.remove("Lima") # elimina el dato indicado | clear() elimina todo
print(mapa)

mapa.sort() # ordena la lista ascendente | procura que todos sean del mismo tipo de datos
print(mapa)

mapa.reverse() # invierte el orden de la lista | procura que todos sean del mismo tipo de datos
print(mapa)

print(mapa.index("4")) # devuelve el número posicional del dato
print(mapa.count("Chicago")) # devuelve la cantidad de veces que se repite el dato
# .copy() copia la lista existente.
```

```python
# método de tuplas, son inmutables no se pueden modificar una vez creadas
comida = ('Ceviche','Torta','Limonada')
print(comida.count('Ceviche')) # cantidad de veces que se repite
print(comida.index('Limonada')) # posición del dato
```

```python
# método de diccionarios clave : valor
libros = {
     "uno":"Calletano",
     'dos':'El señor de los anillos',
     'tres':'Harry Potter'
 }

libros.get('uno') #devuelve el valor según la clasve que se indique

libros.items() #devuelve una tupla por cada (clave:valor)
libros.keys() #devuelve una vista de las claves
libros.values() #devuelve una vista para cada valor

novelas = {
    "cuatro":'El principito',
    'cinco':'Don Quijote'
}

libros.update(novelas) #agrega datos de un diccionario a otro

print(libros.setdefault("tres", "Funca")) #si la clave no existe agrega la clave con el valor que le pongas, si la clave existe te muestra su valor actual
print(libros)

# clear() vacía todo el diccionario
libros.popitem() # borra el último par, clave : valor
print(libros.pop("cuatro")) # borra el par clave:valor por la clave, y devuelve el valor.
print(libros)
```

```python
# método de set (elementos únicos y no ordenados)
animales = {'gato','perro','iguana','loro'}
animales.add('pulga') #agrega datos al set, si es duplicado no lo agrega

figuras = ['cuadrado','circulo','triangulo','cuadrado']
animales.update(figuras) #agrega todo un bloque, ya sea set o lista u otros. Aplica la regla de duplicidad y orden

animales.remove('triangulo') #elimina el dato especificado, da error si no existe

animales.discard("triangulo") #elimina el dato especificado, no da error si no existe

animales.pop() #elimina un dato al azar
animales
```

## Bloque 9
- manejo de errores básicos

```python
try:
    # el código que quieres intentar ejecutar.
    numero = int(input("Introduce un número: "))
    resultado = 10 / numero

except ZeroDivisionError:
    # Qué hacer si ocurre un error específico (ej. dividir entre 0).
    print("¡No puedes dividir entre cero!")

except Exception as e:
    # Un respaldo para cualquier otro error inesperado.
    print(f"Ocurrió un error inesperado: {e}")

finally:
    # Este código se ejecuta SIEMPRE, haya habido error o no.
    # Es ideal para cerrar archivos, conexiones a bases de datos, etc.
    print("Operación finalizada.")
```
